# Зертханалық жұмыс №1
## Қолданбалық деректер жиынын зерттеу: объектілерді, айнымалыларды және талдау мақсатын анықтау

**Пән:** Қосымшалардағы деректерді талдау  
**Деректер жиыны:** `football.csv` — футболшылар туралы мәліметтер  
**Мақсаты:** пәндік саланы сипаттау, объектілер мен айнымалыларды анықтау және нақты талдау мақсатын тұжырымдау.

**Қысқаша теория:** Объект — зерттелетін бірлік; айнымалы — объектінің қасиетін сипаттайтын белгі. Бұл жұмыста әр жол — бір футболшы туралы жазба, әр баған — оның сипаттамасы.

## 1. Кітапханаларды қосу және CSV файлын жүктеу
`football.csv` файлын осы `.ipynb` файлымен бір бумаға орналастырамыз.

```python
import pandas as pd
from IPython.display import display

df = pd.read_csv("football.csv")
print("Жол саны:", df.shape[0])
print("Баған саны:", df.shape[1])
```

## 2. Кемінде 10 жолды қарау

```python
display(df.head(10))
```

## 3. Объект пен айнымалыларды анықтау
**Зерттеу объектісі:** футболшы. **Бақылау бірлігі:** кестедегі бір жол. `Unnamed: 0` — файлдағы жол индексі, ойыншының тұрақты бірегей идентификаторы деп қабылданбайды. `Name` — аты-жөні, қайталануы мүмкін.

## 4. Айнымалыларды түрлерге бөлемыз

```python
variables = pd.DataFrame([
    ["Name", "мәтіндік / атаулық", "L. Messi", "футболшыны атау"],
    ["Age", "сандық", "31", "жас ерекшелігін салыстыру"],
    ["Nationality", "категориялық", "Argentina", "ұлттар бойынша салыстыру"],
    ["Club", "категориялық", "FC Barcelona", "клубтар бойынша топтау"],
    ["Value", "сандық", "110500000", "нарықтық құнын бағалау"],
    ["Wage", "сандық", "565000", "жалақыны салыстыру"],
    ["Position", "категориялық", "RF", "ойын позициясын талдау"],
    ["SprintSpeed", "сандық", "86", "жылдамдық көрсеткішін бағалау"],
], columns=["Айнымалы", "Түрі", "Мән мысалы", "Талдаудағы рөлі"])
display(variables)
```

**Уақыттық айнымалы:** бұл CSV файлында күн/уақытты көрсететін жеке баған жоқ. 


## 5. Деректер типтерін тексеру

```python
selected = ["Name", "Age", "Nationality", "Club", "Value", "Wage", "Position", "SprintSpeed"]
display(pd.DataFrame({"Баған": selected, "pandas типі": [str(df[c].dtype) for c in selected]}))
```

## 6. Зерттеу сұрақтары
1. Футболшылардың жасы мен нарықтық құны арасында қандай байланыс бар?
2. Қай ойын позициясындағы футболшылардың орташа нарықтық құны жоғары?
3. Футболшылардың жылдамдығы (`SprintSpeed`) мен нарықтық құны арасында байланыс байқала ма?

## 7. Негізгі талдау мақсаты
**Футболшылардың жасы, позициясы және жылдамдық көрсеткіштері бойынша нарықтық құнының қалай ерекшеленетінін анықтау.**

## 8. Бастапқы нәтижелерді кестемен көрсету

```python
summary = pd.DataFrame({
    "Көрсеткіш": ["Футболшылар саны", "Бағандар саны", "Орташа жас", "Орташа нарықтық құн", "Клубтар саны"],
    "Мәні": [len(df), len(df.columns), round(df["Age"].mean(), 2), round(df["Value"].mean(), 2), df["Club"].nunique()]
})
display(summary)
```

## 9. Жеке тапсырма: 
Бұл деректер арқылы футболшыларды жасына, клубына, позициясына және ойын қабілеттеріне қарай салыстыруға болады. Нарықтық құн көрсеткіші клубқа ойыншылардың бағасын бағалауға көмектеседі. Жас көрсеткіші әртүрлі жас топтарындағы футболшыларды салыстыруға мүмкіндік береді. `SprintSpeed` арқылы жылдам ойыншыларды іріктеп, олардың нарықтық құнымен байланысын тексеруге болады. Позициялар бойынша орташа құнды есептеу қай рөлдегі ойыншылардың бағасы жоғары екенін көрсетуі мүмкін. Дегенмен нақты трансферлік шешім қабылдау үшін ойыншының қазіргі формасы, жарақаты және деректердің жаңартылған уақыты сияқты қосымша ақпарат қажет.

## 10. Бақылау сұрақтарына жауап
**1. Объект пен айнымалының айырмашылығы қандай?** Объект — зерттелетін бірлік (футболшы), айнымалы — оның белгісі (жасы, клубы).

**2. Категориялық және сандық айнымалылар қалай ажыратылады?** Сандық айнымалылармен арифметикалық есептеулер жасауға болады (`Age`, `Value`), категориялық айнымалылар топтарды білдіреді (`Club`, `Position`).

**3. Талдау мақсатын нақты қою не үшін маңызды?** Ол қажетті көрсеткіштер мен әдістерді таңдауға, нәтижені түсіндіруге және дәлелді шешім қабылдауға көмектеседі.
